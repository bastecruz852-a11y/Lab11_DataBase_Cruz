### 2. Create and Connect to a SQLite Database

```python
import sqlite3
conn = sqlite3.connect("students.db")
cursor = conn.cursor()
print("Connected to students.db")
```

### 3. Create a Table

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    full_name TEXT NOT NULL,
    course TEXT NOT NULL,
    year_level INTEGER NOT NULL
)"""
cursor.execute(create_table_query)
conn.commit()
print("Table 'students' created successfully.")
```

### 5. Query the Data

```python
# All students
cursor.execute("SELECT * FROM students")
print("All:", cursor.fetchall())

# Filter BS ECE
cursor.execute("SELECT * FROM students WHERE course = 'BS ECE'")
print("BS ECE:", cursor.fetchall())

# Sort descending
cursor.execute("SELECT * FROM students ORDER BY year_level DESC")
print("Sorted:", cursor.fetchall())

# Limit top 2
cursor.execute("SELECT * FROM students LIMIT 2")
print("Top 2:", cursor.fetchall())
```

### 12. Create and Join Tables

```python
cursor.execute("CREATE TABLE IF NOT EXISTS courses (course_code TEXT PRIMARY KEY, description TEXT NOT NULL)")
course_data = [("BS ECE", "Electronics Engineering"), ("BS EE", "Electrical Engineering")]
cursor.executemany("INSERT OR IGNORE INTO courses (course_code, description) VALUES (?, ?)", course_data)
conn.commit()

cursor.execute("SELECT s.full_name, c.description FROM students s JOIN courses c ON s.course = c.course_code")
for row in cursor.fetchall():
    print(row)
```

### 12.1 Try It

```python
# Create the courses table [cite: 241, 242]
cursor.execute("""
CREATE TABLE IF NOT EXISTS courses (
    course_code TEXT PRIMARY KEY,
    description TEXT NOT NULL
)
""")
conn.commit()

# Insert predefined course descriptions [cite: 248, 255]
course_data = [
    ("BS ECE", "Electronics and Communications Engineering"),
    ("BS EE", "Electrical Engineering"),
    ("BS COE", "Computer Engineering"),
    ("BS CE", "Civil Engineering")
]
cursor.executemany("INSERT OR IGNORE INTO courses VALUES (?, ?)", course_data)
conn.commit()

# Run the JOIN query to combine both tables [cite: 263, 264]
cursor.execute("""
SELECT s.full_name, s.course, c.description
FROM students s
JOIN courses c ON s.course = c.course_code
""")

print("Student List with Course Descriptions:")
for row in cursor.fetchall():
    print(row)
```

### 13. Create and Join Tables (Filtering/Sorting)

```python
cursor.execute("""SELECT s.full_name, s.course, c.description FROM students s 
                  JOIN courses c ON s.course = c.course_code 
                  WHERE s.course = 'BS CE' ORDER BY s.full_name ASC""")
for row in cursor.fetchall():
    print(row)
```

### 13.1 Try It

```python
# Modified query for BS ECE only, sorted by name [cite: 282]
cursor.execute("""
SELECT s.full_name, s.course, c.description
FROM students s
JOIN courses c ON s.course = c.course_code
WHERE s.course = 'BS ECE'
ORDER BY s.full_name ASC
""")

print("Sorted BS ECE Students:")
results = cursor.fetchall()
if results:
    for row in results:
        print(row)
else:
    print("No BS ECE students found.")
```

### 14. Search for Records Using Pattern Matching

```python
# Parameterized search
keyword = input("Enter a name keyword to search: ")
cursor.execute("SELECT * FROM students WHERE full_name LIKE ?", ('%' + keyword + '%',))
for row in cursor.fetchall():
    print(row)
```
